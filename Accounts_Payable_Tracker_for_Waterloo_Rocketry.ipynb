```python
import pandas as pd

data = '/content/Accounts-Payable-Tracker-for-Waterloo-Rocketry/Transaction Summary'

# Read as tab-separated values
df = pd.read_csv(data, sep='\t', on_bad_lines='skip')

# Clean up column headers by removing any extra quotes, tabs, or trailing commas
df.columns = df.columns.str.replace('\t', '').str.replace(',', '').str.strip()

# Print the clean columns to verify
print("Cleaned Columns:", df.columns.tolist())
print(df.head(2))
```
